## 1. Load the CSV dataset
Let's import pandas and numpy, then load the Combined_dataset.csv file.


```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('Combined_dataset.csv')
print(f'Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns')
```

## 3. Handle missing values
Let's check which columns have missing values first.


### 3.1 Check for nulls


```python
# Null counts per column
nulls = df.isnull().sum()
nulls[nulls > 0]
```

### 3.2 Impute/fill missing values
We will handle the missing values by:
- Imputing rating with the median.
- Filling ratings_count and discount with 0.
- Replacing missing string values in seller, reviews, and description with defaults.


```python
# Fill missing values
df['rating'] = df['rating'].fillna(df['rating'].median())
df['ratings_count'] = df['ratings_count'].fillna(0)
df['discount'] = df['discount'].fillna(0)
df['seller_name'] = df['seller_name'].fillna('Unknown Seller')
df['what_customers_said'] = df['what_customers_said'].fillna('No reviews')
df['product_description'] = df['product_description'].fillna('No description available')

# Check if there are any remaining nulls in these columns
df[['rating', 'ratings_count', 'discount', 'seller_name', 'product_description']].isnull().sum()
```

## 4. Basic operations (filtering and column selection)
Let's filter the data to get items with rating > 4.5 and at least 50 ratings, and view a few columns.


```python
# Filter for highly rated products with substantial feedback
high_rated = df[(df['rating'] > 4.5) & (df['ratings_count'] >= 50)]
print(f'Found {high_rated.shape[0]} matching items')

# Select a subset of columns
cols = ['product_id', 'title', 'rating', 'ratings_count', 'initial_price', 'category']
high_rated[cols].head()
```

## 5. Remove duplicate rows
Let's find duplicates based on product_id and remove them, keeping the first occurrence.


```python
# Count duplicates based on product_id
print(f'Duplicates: {df.duplicated(subset=["product_id"]).sum()}')

# Drop duplicates
df = df.drop_duplicates(subset=['product_id'], keep='first')
print(f'New shape: {df.shape}')
```

## 6. Create derived column (total_amount = price * quantity)
We need to clean the price column first (strip currency symbols and commas, convert to numeric).
Then we'll generate random quantities (1-5) and calculate total_amount.


```python
# Clean price column and convert to float
df['price'] = df['final_price'].astype(str).str.replace('₹', '', regex=False).str.replace(',', '', regex=False).str.replace('"', '', regex=False).str.strip()
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df = df.dropna(subset=['price'])

# Generate mock quantities (1 to 5)
np.random.seed(42)
df['quantity'] = np.random.randint(1, 6, size=len(df))

# Calculate total_amount
df['total_amount'] = df['price'] * df['quantity']

# Print sample values
df[['title', 'price', 'quantity', 'total_amount']].head()
```

## 7. Save the cleaned dataset
Save the cleaned DataFrame as a new CSV file.


```python
# Save as csv
df.to_csv('cleaned_shopping_dataset.csv', index=False)
print('File saved successfully.')
```
